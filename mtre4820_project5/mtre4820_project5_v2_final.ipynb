```python
#Group members: Regina Martinez, Lauren Nunez, Saam Grami

'''
References:
https://www.geeksforgeeks.org/computer-vision/object-detection-with-yolo-and-opencv/
https://docs.ultralytics.com/modes/train/
https://docs.ultralytics.com/models/yolo26/#usage-examples

We created our own models, Lauren was able to trainm them all on her laptop with an NVIDA GPU
'''

import cv2
import random
from ultralytics import YOLO
 
 
model_glasses = YOLO("model/runs/detect/glasses_model/weights/best.pt")
model_headphones = YOLO("model/runs/detect/headphones_model/weights/best.pt")
model_phones = YOLO("model/runs/detect/phones_model/weights/best.pt")

print("Loaded models")
def getColours(cls_num):
    random.seed(cls_num)
    return tuple(random.randint(0, 255) for _ in range(3))
 
cap = cv2.VideoCapture(0)

print("Starting loop")
while True:
    ret, frame = cap.read()
    if not ret:
        continue
 
    
    results_glasses = model_glasses.track(frame, persist=True)
    results_headphones = model_headphones.track(frame, persist=True)
    results_phones = model_phones.track(frame, persist=True)
 
    for results in [results_glasses, results_headphones, results_phones]:
        for result in results:
            class_names = result.names
            for box in result.boxes:
                if box.conf[0] > 0.4:
                    x1, y1, x2, y2 = map(int, box.xyxy[0])
                    cls = int(box.cls[0])
                    class_name = class_names[cls]
                    conf = float(box.conf[0])
                    colour = getColours(cls)
                    cv2.rectangle(frame, (x1, y1), (x2, y2), colour, 2)
                    cv2.putText(frame, f"{class_name} {conf:.2f}",
                                (x1, max(y1 - 10, 20)),
                                cv2.FONT_HERSHEY_SIMPLEX,
                                0.6, colour, 2)
 
    cv2.imshow('YOLO Detection', frame)
    if cv2.waitKey(25) & 0xFF == ord('q'): #Click on video window and press q to clsoe
        break
 
cap.release()
cv2.destroyAllWindows()
```
