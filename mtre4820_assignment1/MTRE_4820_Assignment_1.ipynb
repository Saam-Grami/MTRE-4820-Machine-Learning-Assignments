Cell 3 — Numeric test and plot

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(0, 2*np.pi, 200)
y = np.sin(x)

plt.figure()
plt.plot(x, y)
plt.title("Sanity check: sin(x)")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.show()
print("Plot generated successfully.")
```

Cell 4 — Tiny TensorFlow and PyTorch checks

```python
import tensorflow as tf
import torch
# TensorFlow: simple tensor operation
a = tf.constant([1.0, 2.0, 3.0])
b = tf.constant([4.0, 5.0, 6.0])
print("TensorFlow add:", tf.add(a, b).numpy())
# PyTorch: simple tensor operation
t1 = torch.tensor([1.0, 2.0, 3.0])
t2 = torch.tensor([4.0, 5.0, 6.0])
print("PyTorch add:", (t1 + t2).numpy())
```

Cell 5 — OpenCV check (generate an image and display it)

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
img = np.zeros((120, 240, 3), dtype=np.uint8)
img[:, :120] = (255, 0, 0) # left half (BGR)
img[:, 120:] = (0, 255, 0) # right half (BGR)
plt.figure()
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("OpenCV sanity check: generated image")
plt.axis("off")
plt.show()
print("OpenCV image generated successfully.")
```
